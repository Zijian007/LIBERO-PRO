```python
# Compare New Env vs Original Env
import pathlib
from libero.libero import benchmark
from libero.libero import get_libero_path
from libero.libero.envs import OffScreenRenderEnv
import numpy as np
import matplotlib.pyplot as plt

task_id = 0

def _get_libero_env(task, resolution, seed):
    """Initializes and returns the LIBERO environment, along with the task description."""
    task_description = task.language
    task_bddl_file = pathlib.Path(get_libero_path("bddl_files")) / task.problem_folder / task.bddl_file
    env_args = {"bddl_file_name": task_bddl_file, "camera_heights": resolution, "camera_widths": resolution}
    env = OffScreenRenderEnv(**env_args)
    env.seed(seed)  # IMPORTANT: seed seems to affect object positions even when using fixed initial state
    return env, task_description

# Load New Env (libero_10_temp)
benchmark_dict = benchmark.get_benchmark_dict()
task_suite_new = benchmark_dict[task_suite_name]()
task_new = task_suite_new.get_task(task_id)
env_new, task_description_new = _get_libero_env(task_new, 256, 7)
obs_new = env_new.reset()
img_new = np.ascontiguousarray(obs_new["agentview_image"][::-1, ::-1])

# Load Original Env (libero_10)
task_suite_orig = benchmark_dict["libero_10"]()
task_orig = task_suite_orig.get_task(task_id)
env_orig, task_description_orig = _get_libero_env(task_orig, 256, 7)
obs_orig = env_orig.reset()
img_orig = np.ascontiguousarray(obs_orig["agentview_image"][::-1, ::-1])

# Display both images side by side
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].imshow(img_new)
axes[0].axis('off')
axes[0].set_title(f"New Env\n{task_description_new}")

axes[1].imshow(img_orig)
axes[1].axis('off')
axes[1].set_title(f"Original Env\n{task_description_orig}")

plt.subplots_adjust(wspace=0.3)
plt.tight_layout()
plt.show()
```
